# Build callback

```python

callback = EarlyStoping(
                      monitor="val_loss",
                      mode="min",
                      patience=5,
                      percent_delta=0.1,
                      verbose=True,
                   )
```

```python

logs = {"acc":0.1, "val_acc":0, "loss":0, "val_loss":0}
i = 0
for epoch, log in enumerate(np.random.uniform(low=0.5, high=1.0, size=(50, 4))):
    logs["acc"] = float(log[0])
    logs["val_acc"] = float(log[1])
    logs["loss"] = float(log[2])
    logs["val_loss"] = float(log[3]) + i
    try:
        callback.on_epoch_end(epoch, logs)
    except Exception as e:
        print(e)
        break
```
